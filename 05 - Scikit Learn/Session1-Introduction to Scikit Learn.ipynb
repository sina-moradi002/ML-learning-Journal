#### The Estimator API
At the core of Scikit-learn is the Estimator API, which provides a consistent interface for all machine learning algorithms in the library.

An estimator is any object that learns from data, whether it's a classification, regression, clustering algorithm, or a transformer used for preprocessing.

The Estimator API revolves around three main methods:

1. **fit()**: Learns from the data
2. **predict()**: Makes predictions on new data
3. **transform()**: Modifies the data (for preprocessing and feature engineering)

#### Pipline API
