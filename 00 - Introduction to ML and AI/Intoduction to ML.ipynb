## Introduction to Machine Learning


### Learning vs. Training

#### Learning:
تغییری که باعث می شود مدل از تجربه (داده)، الگو یا دانشی به دست بیاورد.

#### Training:
فرآیندی که ما عمدا انجام می دهیم تا مدل این یادگیری را از روی داده ها انجام دهد.

یعنی به صورت خلاصه:

> $Training$ فرآیندی است که برای ایجاد $Learning$ در مدل انجام می دهیم.

> $Training$ کاری است که انجام می‌دهی؛ $Learning$ تغییری است که در نتیجه‌ی آن در توانایی مدل ایجاد می‌شود.




### What is the *Model* ?

مدل درواقع یک تابع ریاضی است که ورودی را میگیرد و بر اساس پارامتر هایش یک خروجی تولید می کند.
مثلا:
$$ y = wx + b$$

این یک مدل بسیار ساده است که در آن:
- $x$ ورودی
- $y$ خروجی
- $w$ و $b$ پارامتر های مدل
است.

پس:
> $Model$ ساختاری که رابطه بین $input$ و $output$ را نمایش می دهد.

هدف در فرآیند آموزش ($Training$) پیدا کردن بهترین مقادیر برای پارامتر های مدل است



### What is *Machine Learning* ?

در برنامه نویسی معمولی به کامپیوتر می گوییم:
```
Input + Rules -> Output
```

اما در ماشین لرنینگ میخواهیم کامپیوتر خودش یک رابطه (قاعده) را از *داده ها* پیدا کند:
```
Input + Output examples
          ↓
    Machine Learning
          ↓
        Model
```

درواقع:
> $Machine Learning$ مجموعه ای از روش ها برای ساختن مدل هایی است که بتوانند از داده الگو یاد بگیرند و بر اساس آن روی داده های جدید پیش بینی یا تصمیم گیری کنند.

---

#### Goal:
هدف در یادگیری ماشین معمولا این نیست که مدل روی داده های آموزشی جواب های درست بدهد

هدف اصلی این است که مدل روی داده هایی که قبلا ندیده است عملکرد خوبی داشته باشد
به این میگوییم:

**Generalization**

برای مثال، فرض کنید 1000 عکس مختلف از سگ ها و گربه ها داریم.

800 عکس را برای آموزش داده استفاده می کنیم

بعد 200 عکس را به مدل نشان می دهیم که قبلا ندیده است.

اگر مدل روی این 200 عکس هم خوب عمل کند یعنی $generalization$ خوبی دارد.

---

#### Kinds of Machine Learning

- machine learning

### Loss Function
باید راهی داشته باشیم که بفهمیم مدل چقدر خوب یا چقدر بد است!
اینجا از این تابع استفاده می کنیم.

درواقعیت آنها توابع ریاضیاتی دقیق هستند مانند $MSE$

هدف آموزش این است که پارامتر ها را طوری پیدا کنیم که این مقدار کمترین حد ممکن شود



### Optimization
فرض کنید مدلی به شکل زیر داریم:
$$ y = wx + b $$
ما مقدار $b$ و $w$ را نمیدانیم

ابتدا مقادیری را برای آنها در نظر گرفته ایم ولی مقدار $Loss$ زیاد است.

حالا مسئله این است:
> چطور $b$ و $w$ را تغییر دهیم که که مقدار $Loss$ کمتر شود؟؟

این مسئله یعنی:
**Optimization**

یعنی پیدا کردن پارامتر هایی که یک تابع هدف را کمینه یا بیشینه می کنند

متد های زیادی برای این کار وجود دارد که بسته به شرایط باید به کار گرفته شوند.



### Machine Leaning Overview
```
                    MACHINE LEARNING
                           │
                           ↓
                         DATA
                           │
                           ↓
                         MODEL
                        f(x; θ)
                           │
                           ↓
                       TRAINING
                           │
               ┌───────────┴───────────┐
               ↓                       ↓
          Prediction                 Truth
               │                       │
               └───────────┬───────────┘
                           ↓
                          LOSS
                           │
                           ↓
                     OPTIMIZATION
                           │
                     Gradient Descent
                           │
                           ↓
                    Update Parameters
                           │
                           └──────→ Repeat
                                      │
                                      ↓
                              TRAINED MODEL
                                      │
                                      ↓
                                  EVALUATION
                                      │
                                      ↓
                                  INFERENCE
                                 (Deployment)
```

### Key words

| مفهوم                | معنی خیلی ساده                                        |
| -------------------- | ----------------------------------------------------- |
| **Machine Learning** | یادگیری الگو از داده                                  |
| **Model**            | ساختاری که الگو را نمایش می‌دهد                       |
| **Training**         | فرایند تنظیم مدل با داده                              |
| **Learning**         | تغییر/کسب پارامترها یا الگو از تجربه‌ی داده‌ای        |
| **Loss**             | اندازه‌گیری میزان اشتباه                              |
| **Optimization**     | پیدا کردن پارامترهای بهتر برای کاهش Loss              |
| **Estimator**        | چیزی که کمیت/پارامتر ناشناخته را از داده تخمین می‌زند |
