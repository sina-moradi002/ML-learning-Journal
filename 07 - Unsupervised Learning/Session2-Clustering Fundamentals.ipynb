یک ابزار قدرتمند برای کشف الگوها و ساختارهای پنهان در مجموعه داده‌های پیچیده است

Clustering is the process of partitioning a set of data points into groups, called clusters, based on their similarities. In essence, it's about:

- **Grouping:** Organizing data points into meaningful groups.
- **Similarity:** Ensuring that points within a cluster are more similar to each other than to points in other clusters.
- **Unsupervised Learning:** Discovering patterns without pre-labeled data.

![](images/img_2.png)

Here are some key concepts in clustering:

- **Clusters:** Groups of data points that share similar characteristics.
- **Centroids:** The center point of a cluster, often used in algorithms like K-means.
- **Intra-cluster distance:** The distance between points within the same cluster.
- **Inter-cluster distance:** The distance between points in different clusters.

The primary goal of clustering is to maximize intra-cluster similarity while minimizing inter-cluster similarity. Mathematically, we can express this as:
$$\text{Maximize } \sum_{i=1}^{k} \sum_{x \in C_i} \text{similarity}(x, c_i)$$

Where $k$ is the number of clusters, $C_i$ is the $i$-th cluster, $x$ is a data point in cluster$C_i$, and $c_i$ is the centroid of cluster $C_i$.

#### Internal Evaluation Metrics
این روش ها متکی بر اطلاعات داخلی داده ها هستند و صرفا بر اساس آنچه از داده ها میبینیم اقدام به ارزیابی مدل می کنند

1. **Silhouette Score**
    - بررسی می کند که هر آبجکت چقدر به سایر آبجکت های درون خوشه خود شباهت دارد
    - Range: [-1, 1]
    - Higher values indicate better-defined clusters

2. **Calinski-Harabasz Index**
    - نسبت پراکندگی بین خوشه‌ای و پراکندگی درون خوشه‌ای
    - مقادیر بالاتر نشان دهنده خوشه‌های بهتر هستند

3. **Davies-Bouldin Index**
    - میانگین «شباهت» بین هر خوشه و شبیه‌ترین خوشه به آن
    - شباهت، نسبت فواصل درون خوشه‌ای به فواصل بین خوشه‌ای است.
    - Lower values indicate better clustering
