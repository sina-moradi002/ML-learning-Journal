## **Session 1** -Introduction to Regression
رگرسیون یک مفهوم اساسی در یادگیری ماشین و آمار است که بر درک و کمی‌سازی رابطه بین متغیرها تمرکز دارد. در اصل، تحلیل رگرسیون با هدف مدل‌سازی چگونگی تأثیر یک یا چند متغیر مستقل بر یک متغیر وابسته انجام می‌شود.

Regression is a fundamental concept in machine learning and statistics that focuses on understanding and quantifying the relationship between variables. At its core, regression analysis aims to model how one or more independent variables influence a dependent variable.

In its simplest form, a regression model can be represented as:
$$y = f(X) + \epsilon$$

where:
- $y$ is the dependent variable
- $X$ represents one or more independent variables
- $f$ is the function that describes the relationship between $X$ and $y$
- $\epsilon$ is the error term, representing the deviation from the true relationship



### Importance and Applications
تحلیل رگرسیون یک روش آماری قدرتمند است که نقش حیاتی در زمینه‌های مختلف مطالعاتی و کاربردهای دنیای واقعی ایفا می‌کند. اهمیت آن ناشی از توانایی آن در کشف روابط بین متغیرها، انجام پیش‌بینی‌ها و اطلاع‌رسانی در فرآیندهای تصمیم‌گیری است.

why regression is essential:

1. **Predictive Modeling:** Regression allows us to forecast future outcomes based on historical data.
2. **Relationship Quantification:** It helps quantify the strength and nature of relationships between variables.
3. **Hypothesis Testing:** Regression can be used to test scientific hypotheses about the relationships between variables.
4. **Feature Selection:** It aids in identifying which independent variables have the most significant impact on the dependent variable.
5. **Decision Support:** Regression models provide valuable insights that can guide business and policy decisions.

---

#### Example: House price

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

# Generate data
np.random.seed(42)
x = np.random.rand(100, 1)*10
y = 2 * x + 1 + 3 * np.random.randn(100, 1)

# Split data to train and test
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Create and train the linear regression model
model =LinearRegression()
model.fit(X_train, y_train)

# Generate predictions for the entire range of x
X_range = np.linspace(0, 10, 100).reshape(-1, 1)
y_pred = model.predict(X_range)

# Create Scatterplot
plt.figure(figsize=(10, 6))
plt.scatter(X_train, y_train, color='blue', label='Training data')
plt.scatter(X_test, y_test, color='green', label='Test data')
plt.plot(X_range, y_pred, color='red', label='Regression Line')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression Example')
plt.legend()
plt.grid(True)
plt.show()
```

### Types of Regression Problems

#### Simple vs. Multiple Regression

رگرسیون ساده یعنی فقط به یک متغیر وابسته است. درواقع یعنی تابع آن یک متغیره است

**Definition:** $y = f(x) + \epsilon$

رگرسیون چند گانه یعنی تابع آن چند متغییره است و به چند متغیر وابسته است

**Definition:** $y = f(x_1, x_2, ..., x_n) + \epsilon$

---

#### Linear vs. Non-Linear Regression

**Linear:**
در این رگرسیون تابع ، یک تابع خطی است

- *simple linear:* $y = \beta_0 + \beta_1x + \epsilon$
- *Multiple Linear:* $y = \beta_0 + \beta_1x_1 + \beta_2x_2 + ... + \beta_nx_n + \epsilon$

```python
# Linear Regression
x = np.linspace(0, 5, 100)
y = 2*x + np.random.normal(0, 1, 100)

plt.figure(figsize=(8 , 6))
plt.scatter(x, y)
plt.plot(x, 2*x, color='red', label='True Relationship')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression Example')
plt.legend()
plt.grid(True)
plt.show()
```

**Non_Linear regression:**
در این نوع تابع رگرسیون یک تابع غیر خطی است

*Examples:* Polynomial regression, exponential regression, logistic regression

*form:* $y = f(X, \beta) + \epsilon$


```python
x = np.linspace(0, 5, 100)
y = 2*x**2 + np.random.normal(0, 5, 100)

plt.figure(figsize=(8, 6))
plt.scatter(x, y)
plt.plot(x, 2*x**2, color='red', label='True Relationship')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Non-Linear Regression Example')
plt.legend()
plt.grid(True)
plt.show()
```

#### Other Types of Regression

1. **Logistic Regression:** Used for binary classification problems, but technically a regression method.
2. **Ridge Regression:** Linear regression with L2 regularization to prevent overfitting.
3. **Lasso Regression:** Linear regression with L1 regularization, useful for feature selection.
4. **Elastic Net:** Combines both L1 and L2 regularization.
5. **Polynomial Regression:** Fits a non-linear relationship using polynomial terms.
6. **Stepwise Regression:** Iteratively adds or removes predictors to find the best model.



### Key Concepts of Regression
#### Dependent and independent Variables:

**Dependent Variable (Y):**

- Also known as the response variable, outcome variable, or target variable.
- The variable we're trying to predict or explain.
- Usually plotted on the vertical (y) axis.

**Independent Variables (X):**

- Also known as predictor variables, explanatory variables, or features.
- The variables used to predict or explain the dependent variable.
- Usually plotted on the horizontal (x) axis in simple regression.

---

#### Regression Line/Curve

The regression line (or curve for non-linear regression) is the mathematical representation of the relationship between the dependent and independent variables.

For simple linear regression: $Y = \beta_0 + \beta_1X$

---

#### Residuals and Error

Residuals are the differences between the observed values and the predicted values from the regression model.

- $Residual = Observed value - Predicted value$
- The goal in regression is to minimize these residuals.

---

**Visualizing Residuals:**

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import seaborn as sns

# Set random seed for reproducibility
np.random.seed(42)

# Generate sample data
X = np.linspace(0, 10, 100).reshape(-1, 1)
y = 2 * X + 1 + np.random.normal(0, 1, (100, 1))

# Perform linear regression
model = LinearRegression()
model.fit(X, y)

# Make predictions
y_pred = model.predict(X)

# Calculate residuals
residuals = y - y_pred

# Set up the plot style
sns.set_palette("husl")

# Create the main figure and subplots
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 10), gridspec_kw={'height_ratios': [2, 1]})

# Plot 1: Scatter plot with regression line
ax1.scatter(X, y, alpha=0.7, s=50, label='Data points')
ax1.plot(X, y_pred, color='red', linewidth=2, label='Regression line')

# Add labels and title
ax1.set_xlabel('X', fontsize=14)
ax1.set_ylabel('Y', fontsize=14)
ax1.set_title('Linear Regression with Residuals', fontsize=18, fontweight='bold')
ax1.legend(fontsize=12)

# Annotate the model equation
equation = f'y = {model.coef_[0][0]:.2f}x + {model.intercept_[0]:.2f}'
mse = mean_squared_error(y, y_pred)
ax1.text(
    0.05, 0.95, f'Model: {equation}\nMSE: {mse:.4f}', transform=ax1.transAxes,
    fontsize=12, verticalalignment='top', bbox=dict(boxstyle='round', facecolor='white', alpha=0.7)
)

# Plot residuals
for i in range(len(X)):
    ax1.plot([X[i], X[i]], [y[i], y_pred[i]], 'r-', linewidth=1, alpha=0.7)

# Plot 2: Residual plot
ax2.scatter(X, residuals, alpha=0.7, s=50)
ax2.axhline(y=0, color='green', linestyle='--', linewidth=2)

# Add labels and title for residual plot
ax2.set_xlabel('X', fontsize=14)
ax2.set_ylabel('Residuals', fontsize=14)
ax2.set_title('Residual Plot', fontsize=16, fontweight='bold')

# Adjust layout and display the plot
plt.tight_layout()
plt.show()
```

### Steps in a typical regression analysis

#### Problem Definition
Start by clearly defining the problem you're trying to solve:

- Identify the dependent variable (what you want to predict)
- Determine potential independent variables (predictors)
- Formulate your research question or hypothesis
- Example: "Can we predict a house's price based on its size, location, and age?"

---

#### Data Collection
Gather relevant data for your analysis:

- Ensure data quality and reliability
- Collect sufficient samples to represent the population
- Consider both historical and current data if applicable

---

#### Data Preprocessing
Prepare your data for analysis:

- Handle missing values
- Detect and treat outliers
- Encode categorical variables
- Normalize or standardize features if necessary

---

#### Exploratory Data Analysis (EDA)
Explore your data to gain insights:

- Visualize relationships between variables
- Check for correlations
- Identify potential patterns or trends

---

#### Model Selection
Choose an appropriate regression model based on your data and problem:

- Linear Regression for linear relationships
- Polynomial Regression for non-linear patterns
- Multiple Regression for multiple predictors

---

#### Model Training
Fit the selected model to your data:

- Split data into training and testing sets
- Train the model on the training data

---

#### Model Evaluation
Assess the performance of your model:

- Use appropriate metrics (e.g., R-squared, MSE, RMSE)
- Validate on the test set
- Perform cross-validation if necessary

---

#### Model Diagnostics
Check if the model meets regression assumptions:

- Linearity
- Independence
- Homoscedasticity
- Normality of residuals

---

#### Model Refinement
If necessary, refine your model:

- Feature selection or engineering
- Try different algorithms
- Adjust hyperparameters
- Address any violations of assumptions

---

#### Interpretation and Reporting
Interpret the results and communicate findings:

- Explain the relationship between variables
- Discuss the model's predictive power
- Address limitations and potential improvements

---

#### Deployment and Monitoring (if applicable)
If the model is to be used in production:

- Deploy the model in the intended environment
- Set up monitoring to track performance over time
- Plan for periodic retraining or updates
