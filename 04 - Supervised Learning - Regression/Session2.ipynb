## **Session 2** - Simple Linear Regression
رگسرسیون ساده خطی در واقع دارای تابعی است که تنها یک متغیر دارد.

این به معنی است که یک متغیر وابسته $y$ به تنها یک متغیر غیر وابسته (مستقل) $x$ ، وابسته است.

Formulation: $$y = \beta_0 + \beta_1x + \epsilon$$

- $y$: The dependent variable (outcome or target)
- $x$: The independent variable (predictor or feature)
- $\beta_0$: The y-intercept (the value of $y$ when x = 0)
- $\beta_1$: The slope (how much y changes for a one-unit increase in x)
- $\epsilon$: The error term (captures all other factors that influence y)




### The linear Regression Model
The Linear Regression Model forms the core of simple linear regression analysis. It provides a mathematical framework for understanding and predicting the relationship between two variables.

 The linear regression model assumes that the relationship between the independent variable $(x)$ and the dependent variable $(y)$ can be represented by a straight line.

### Assumptions of Simple Linear Regression
1. Linearity

    تغییرات متغیر $y$ بر حسب $x$ باید خطی باشد. یعنی نسبت $y/x$ در هر نقطه ثابت باشد.

2. Independence

    مشاهدات باید مستقل از یکدیگر باشند. این بدان معناست که مقدار یک مشاهده نباید بر مشاهدات دیگر تأثیر بگذارد یا از آنها تأثیر بپذیرد.

    درواقع این یعنی مقادیر ارور ها نباید نظم خاصی داشته باشد و هرگونه تاثیر آن ها به هم نقض قاعده اسقلال است.

3. Homoscedasticity

    واریانس باقیمانده‌ها باید در تمام سطوح متغیر مستقل ثابت باشد. این فرض همچنین به عنوان همگنی واریانس شناخته می‌شود.

    به زبان ساده یعنی، مقادیر ارور ها باید در بازه های مختلف از تابع تقریبا ثابت باشد . یک جا مقادیر ارور خیلی زیاد و یک جا خیلی کم نباشد

4. Normality

    باقیمانده‌ها (خطاها) باید توزیع نرمال داشته باشند.

5. No or Little Multicollinearity
6. No Outliers

---

#### Checking Assumptions

To validate these assumptions, you can use various techniques:

1. **Scatter plots:** To check linearity and homoscedasticity
2. **Q-Q plots:** To check normality of residuals
3. **Residual plots:** To check linearity, homoscedasticity, and identify outliers
4. **Durbin-Watson test:** To check independence


### Maximum Likelihood Formulation

this is the Maximum Likelihood Estimation (MLE) approach for linear regression.

در رگرسیون خطی، معمولاً فرض می‌کنیم که $y$ از توزیع نرمال با میانگینی که به $x$ و واریانس ثابت بستگی دارد، پیروی می‌کند.

Mathematically, we can express this as: $$y_i \sim \mathcal{N}(\mu_i, \sigma^2)$$
where: $\mu_i = \beta_0 + \beta_1x_i$
Here:
- $\mathcal{N}$ توزیع نرمال را نشان میدهد
- $\mu_i$ میانگین مشاهده $i$ ام را نشان میدهد
- $\sigma^2$ واریانس خط ها را نشان میدهد

---

#### Likelihood Function
The likelihood function is the joint probability of observing our data, given the model parameters. For $n$ independent observations, the likelihood function is:

$L(\beta_0, \beta_1| \mathbf{x}, \mathbf{y}) = \prod_{i=1}^n \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(y_i - (\beta_0 + \beta_1x_i))^2}{2\sigma^2}\right)$

---

#### Log(Likelihood)
برای سادگی در محاسبات لگاریتم میگیریم

$\ell(\beta_0, \beta_1 | \mathbf{x}, \mathbf{y}) = -\frac{n}{2}\log(2\pi\sigma^2) - \frac{1}{2\sigma^2}\sum_{i=1}^n (y_i - (\beta_0 + \beta_1x_i))^2$

---

#### Maximum Likelihood Estimation
هدف ما طبعا پیدا کردن مقادیر مناسب برای $\beta_0$ و $\beta_1$ و $\sigma^2$ است که تابع لگاریتم مرحله قبل را ماکسیمایز کند.

لذا باید از تابع مورد نظر مشتق جزئی گرفته و برابر صفر قرار دهیم تا بهترین مقادیر بدست آید

$\frac{\partial \ell}{\partial \beta_0} = 0$

$\frac{\partial \ell}{\partial \beta_1} = 0$

Solving these equations leads to the following estimators:

$\hat{\beta_1} = \frac{\sum_{i=1}^n (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^n (x_i - \bar{x})^2}$

$\hat{\beta_0} = \bar{y} - \hat{\beta_1}\bar{x}$



### Ordinary Least Squares (OLS) Method
ایده اصلی $OLS$، به حداقل رساندن مجموع مربعات اختلاف بین مقادیر مشاهده شده و مقادیر پیش‌بینی شده از مدل است.

The core idea behind OLS is to minimize the sum of the squared differences between the observed values and the predicted values from the model.

Mathematically, we aim to minimize:
$$\sum_{i=1}^n (y_i - (\beta_0 + \beta_1x_i))^2$$

Where:

 - $y_i$ is the observed value
 - $\beta_0 + \beta_1x_i$ is the predicted value

By minimizing this sum of squared residuals, we obtain the following estimators:

$\hat{\beta_1} = \frac{\sum_{i=1}^n (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^n (x_i - \bar{x})^2}$

$\hat{\beta_0} = \bar{y} - \hat{\beta_1}\bar{x}$

- $\bar{y}$ and $\bar{x}$ are means



### Visualizing Simple Linear Regression
#### Scatter Plot with Regression Line
این نمودار ساده ترین نمودار برای نمایش رگرسیون خطی ساده است. این نمودار داده های بدست آمده را به صورت یک نمودار نقطه ای نشان میدهد که در اطراف خط رگرسیون پخش شده اند.


```python
from matplotlib.lines import lineStyles
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

#Generate sample data
np.random.seed(42)
X = np.linspace(0, 10 , 100).reshape(-1, 1)
y = 2* X + np.random.normal(0, 1, X.shape) + 1 #Errors should be Normal Distributed

# Fit the model
model = LinearRegression()
model.fit(X, y)

#Plot
plt.figure(figsize = (10,6))
plt.scatter(X, y, color = 'blue', alpha = 0.7)
plt.plot(X, model.predict(X), color = 'red', linewidth = 2)
plt.xlabel('Independent Variable (x)')
plt.ylabel('Dependent Variable (y)')
plt.title('Scatter Plot with Regression Line')
plt.show()
```

#### Residual Plot
این نمودار مقادیر و پراکندگی مقدار های ارور را در اطراف صفر نشان میدهد.

با این نمودار می توانیم بفهمیم که ارور های ما آیا از الگوی خاصی پیروی می کنند یا نه؟ نظم خاصی دارند؟

در یک مدل برازش یافته، باقیمانده‌ها باید به طور تصادفی و بدون هیچ الگوی مشخصی در اطراف خط صفر پراکنده شده باشند.



```python
residuals = y - model.predict(X)
plt.figure(figsize = (10,6))
plt.scatter(X, residuals, color = 'blue', alpha = 0.7)
plt.axhline(y = 0, color = 'red', linestyle = '--')
plt.xlabel('Independent Variable (x)')
plt.ylabel('Residuals')
plt.title('Residuals Plot')
plt.show()
```

#### Q-Q Plot
A Quantile-Quantile (Q-Q) plot helps assess whether the residuals follow a normal distribution.

این نمودار، به ارزیابی اینکه آیا باقیمانده‌ها از توزیع نرمال پیروی می‌کنند یا خیر، کمک می‌کند.



```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Assuming you have already calculated the residuals
residuals = y - model.predict(X)

fig, ax = plt.subplots(figsize=(10, 6))

# Calculate the quantiles
stats.probplot(residuals.flatten(), dist="norm", plot=ax);
```
