## **Session 3** - Expectation, Variance, Moments

### Expected value
Expected value = mean , average
for discrete values:
$$ E[X] = \sum_{x} x \cdot p(x) $$
for continuous values:
$$ E[X] = \int_{-\infty}^{\infty} x \cdot f(x) dx $$


### Variance
for random variable X with expected value E[X] Variance is:
$$ Var(X) = \sum_{x}[(X - E[X])^2] $$
for discrete variables:
$$ Var(X) = \sum_x (x - E[X])^2 \cdot p(x) $$
for continuous variables:
$$ Var(X) = \int_{-\infty}^{\infty} (x - E[X])^2 \cdot f(x) dx $$

Standard deviation:
$$ \sigma = \sqrt{Var(X)} $$


### Covariance and Correlation
#### Covariance:
Covariance measures how two variables change together. It tells us about the direction of the linear relationship between variables.
$$ Cov(X,Y) = E[(X - E[X])(Y - E[Y])] $$
#### ** Correlation: **
The correlation coefficient standardizes covariance, giving us a measure that's easier to interpret. It's defined as:
$$ \rho_{X,Y} = \frac{Cov(X,Y)}{\sigma_X \sigma_Y} $$

### Moments
$ «Moments» $ یا «گشتاورها» در آمار، ابزارهایی هستند برای این‌که بفهمیم یک توزیع داده‌ها چه شکلی دارد؛ یعنی داده‌ها معمولاً کجا جمع شده‌اند، چقدر پخش شده‌اند، چپ‌کج‌اند یا راست‌کج، و چقدر نوک‌تیز یا پهن‌اند.

اگر یک مجموعه داده را مثل یک شکل روی محور اعداد تصور کنیم، $moments$ به ما کمک می‌کنند آن شکل را با چند عدد خلاصه کنیم.

به زبان خیلی ساده:

- **مومنت اول:** محل مرکز یا میانگین را نشان می دهد.
- **مومنت دوم:** میزان پراکندگی را نشان می دهد.
- **مومنت سوم:** کجی توزیع را نشان می دهد.
- **مومنت چهارم:** پهنی یا نوک تیزی را نشان می دهد.

در آمار $moments$ از یک متغیر تصادفی $X$ معمولا به صورت یک میانگین از توان های آن متغیر تعریف می شود.

#### مومنت اول: میانگین
$$ E[X] $$

#### مومنت دوم مرکزی: واریانس
$$ E[(X - \mu)^2] $$

هرچه واریانس داده ها بزرک تر باشد داده ها پخش ترند

#### مومنت سوم مرکزی: $Skewness$ یا چولگی
$$ E[(X - \mu)^3] $$
این مومنت نشان میدهد توزیع به کدام سمت کج شده است.

- اگر مثبت باشد: دم توزیع به سمت راست بلند تر است
- اگر منفی باشد: دم توزیع به سمت چپ بلند تر است
- اگر نزدیک به صفر باشد: توزیع تقریبا متقارن است.

#### مومنت چهارم: $Kurtosis$ یا کشیدگی / نوک تیزی
$$ E[(X - \mu)^4] $$
